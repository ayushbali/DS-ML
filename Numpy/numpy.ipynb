### Arithmetic

````python
a1 = np.array([1,2,3])
a2 = np.array([[1,2,3],[4,5,6]])

# Broadcasting is a powerful feature of NumPy that allows you to perform operations on arrays of different shapes and sizes. It automatically expands the smaller array to match the shape of the larger array, allowing for element-wise operations without the need for explicit (without) looping.

print(a1 * a2) # the multiplication is done element-wise and smaller array is broadcasted to match the shape of the larger array. In this case, a1 is broadcasted to match the shape of a2, and then the multiplication is performed element-wise.
print("-"*100)

print(a1+a2)
print("-"*100)

print(np.log2(a1))
print("-"*100)

# Broadcasting with 2D arrays
a3 = np.array([[1,2,3],[4,5,6]])
a4 = np.array([[1],[2]])
print(a3 + a4) # the smaller array a4 is broadcasted to match the
````

### Reshaping and Transposing
- Rules of Broadcasting 
- two dimensions are equal if
- they have same shapes
- if not one of them has to be 1 

````python
# Reshape
arr2 = np.random.randint(0, 10, size=(2,3))
arr3 = np.random.randint(0, 10, size=(2,3,4))
print(arr2)
print(arr3)

# arr2 * arr3  #! Error because the shapes of the arrays are not compatible for broadcasting. 

(arr2.reshape(2,3,1)) * arr3 # Broadcasting is now possible. because arr2 is now reshaped to (2,3,1) which is compatible with arr3's shape of (2,3,4). The 1 in the last dimension of arr2 allows it to be broadcasted across the 4 columns of arr3.

# Transpose
arr2.shape
print(arr2.T.shape) # Transpose of arr2 changes its shape from (2,3) to (3,2)
print(arr3.T.shape) # Transpose of arr3 changes its shape from (2,3,4) to (4,3,2)
````

### Sorting Arrays

- argsort , argmin, argmax 

````python
np.random.seed(0)
random_array = np.random.randint(0,10, size= (3,4))
print('Original Array:')
print(random_array)
print('-'*100)

print('Sorted Array:')
print(np.sort(random_array))
print('-'*100)

print('Sorted Array along axis 1 (cols):')
print(np.sort(random_array, axis=1)) # sort along the cols
print('-'*100)

print('Sorted Array along axis 0 (rows):')
print(np.sort(random_array, axis=0)) # sort along the rows
print('-'*100)

# Argsort
print('Argsort Array:')
print(np.argsort(random_array)) # returns the indices that would sort the array along the specified axis. By default, it sorts along the last axis (axis=-1).
````
