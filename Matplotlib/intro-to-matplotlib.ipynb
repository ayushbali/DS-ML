## Introduction to Matplotlib

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x_data = np.arange(0, 10)
y_data = np.arange(10,20)
plt.plot(x_data, y_data);
```

```python
# Method 1
fig = plt.figure() 
ax = fig.add_subplot() # add a subplot to the figure
ax.plot(x_data, y_data);
plt.show() # show the figure
```

```python
# Method 2
fig = plt.figure()
ax = fig.add_axes([0,0,1,1]) # add axes to the figure
ax.plot(x_data, y_data);
plt.show() # show the figure
```

```python
# Method 3 (Recommended) 
fig, ax = plt.subplots() # create a figure and a set of subplots
ax.plot(x_data, y_data);
```

### Example Workflow

```python
# import
import matplotlib.pyplot as plt

# prepare data
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# Setup plot - create a figure and axis
fig, ax = plt.subplots(figsize=(5, 5)) # create a figure and a set of subplots with a specific size (x, y) in inches

# Plot data
ax.plot(x,y, color='blue', linewidth=1, linestyle='-', marker='o', markersize=8, markerfacecolor='red', markeredgewidth=0.2, markeredgecolor='black')

# Add title and labels
ax.set(title= 'First Plot', xlabel='x-axis', ylabel='y-axis')

# Add grid
ax.grid()

# Save the figure
# plt.savefig('./images/first_plot.png', dpi=300) # save the figure with a specific resolution
```

```python
sigmoid = lambda x: 1/(1+np.exp(-x)) # sigmoid function
relu = lambda x: np.maximum(0, x) # relu function

x = np.linspace(-10, 10, 10) # create an array of 10 values between -10 and 10
y = sigmoid(x) # apply the sigmoid function to the array

x1 = np.linspace(-10, 10, 10) # create an array of 10 values between -10 and 10
y1 = relu(x1) # apply the relu function to the array

# Setup plot - create a figure and axis
fig, ax = plt.subplots(figsize=(5, 5)) 
ax.grid()
ax.plot(x,y, color='blue', linewidth=1, linestyle='-')
ax.plot(x1,y1, color='red', linewidth=1, linestyle='-')

# Scatter Plot
ax.scatter(x,y, color='blue', s=50, marker='o', alpha=0.5, label='sigmoid') # s is the size of the marker, alpha is the transparency, label is the legend

# Bar Plot
x2 = ['A', 'B', 'C', 'D', 'E']
y2 = np.linspace(1, 10, 5) # create an array of 5 values between 1 and 10

ax.bar(x2, y2, color='orange', alpha=0.7, label='bar plot')


```

```python
# Plot from dictionary
butter_prices = {
    "almond": 2.99,
    "peanut": 1.99,
    "cashew": 3.99,
}
fig, ax = plt.subplots(figsize=(5, 5))
ax.grid()
ax.set(title= 'Butter Prices', xlabel='Butter Type', ylabel='Price ($)')
x = butter_prices.keys()
y = butter_prices.values()

print(type(x), type(y)) # dict_keys, dict_values
ax.bar(x, y, color='green', alpha=0.5, label='bar') # histogram of the values in the dictionary
```

```python
# Histogram
data = np.random.randn(1000) # create an array of 1000 random values
fig, ax = plt.subplots(figsize=(5, 5))
ax.hist(data, bins=30, color='purple', alpha=0.5, label='histogram') # histogram of the values in the array
```

### Multiple Plots

```python
# Method 1 - Subplots
np.random.seed(0) # set the seed for reproducibility
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(10,10))

data = np.random.randn(500) # create an array of 1000 random values

# print(data)

# Plot 1 - Scatter Plot
ax1.set(title='Scatter Plot', xlabel= 'mouse-size', ylabel='mouse-height')
ax1.scatter(data, np.exp(data), color='red')

# Plot 2 - Bar Plot
ax2.set(title='Bar Plot', xlabel= 'mouse-size', ylabel='mouse-height')
ax2.bar(data, sigmoid(data), color='blue')

# Plot 3 - Histogram
ax3.set(title='Histogram', xlabel= 'mouse-size', ylabel='mouse-height')
ax3.hist(data, bins=30, color='green', alpha=0.5)

# Plot 4 - Line Plot
ax4.set(title='Line Plot', xlabel= 'mouse-size', ylabel='mouse-height')
ax4.plot(data, relu(data), color='orange', linewidth=1)
```

```python
# Method 2 - Subplots 
np.random.seed(0) # set the seed for reproducibility
x = np.random.randint(1, 10, 100) # create an array of 100 random integers between 1 and 10
y = np.random.randint(1, 10, 100) # create an array of

fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10,10))
ax[0,0].plot(x,x/2)
ax[0,1].scatter(x,np.sin(x))
ax[1,0].barh(x,x*2)
ax[1,1].hist(x, bins=30, color='purple', alpha=0.5, label='histogram') # histogram of the values in the array
```

## Plotting from Pandas

```python
# Plotting from a pandas Series
# Time seies data is a common use case for pandas, and matplotlib can easily plot this data.
date_range = pd.date_range('2026-01-01', '2026-09-06')

sales_overtime = pd.Series(np.random.randn(len(date_range)), index=date_range)
sales_overtime = sales_overtime.cumsum()
sales_overtime.plot(figsize=(10,5), title='Sales Overtime', xlabel='Date', ylabel='Sales ($)', color='green', alpha=0.5, grid=True)
```
